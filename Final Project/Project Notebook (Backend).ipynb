## Outlier detection for hourly data data using Boxplot

```python
# Define continuous variables to check for outliers

cols_to_check = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']

# Visualizing Outliers for Hourly Data
plt.figure(figsize=(15, 10))
for i, col in enumerate(cols_to_check):
    plt.subplot(2, 3, i + 1)
    sns.boxplot(x=data_h[col])
    plt.title(f'Hourly: {col}')
plt.tight_layout()
plt.show()
```

```python
# Function to remove outliers from datasets

def remove_outliers_iqr(df, columns):
   
    df_clean = df.copy()
    
    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Filter the data: Keep rows within bounds
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]
        
    return df_clean


target_cols = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
```

```python
# Remove outliers

data_h_clean = remove_outliers_iqr(data_h, target_cols)

# Verify Results

print(f"Original Rows: {data_h.shape[0]}")
print(f"Cleaned Rows:  {data_h_clean.shape[0]}")
print(f"Rows Dropped:  {data_h.shape[0] - data_h_clean.shape[0]}")
```

## Model Training

#### Model training for hourly data

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor 
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.metrics import mean_absolute_error

# Define predictors (X) and target (y)
X_h = data_h_clean.drop(['cnt'], axis=1)
y_h = data_h_clean['cnt']
```

```python
# Split into training and testing sets
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_h, y_h, test_size=0.2, random_state=42)
```

```python
# Initialize Random Forest Regressor
rf_h = RandomForestRegressor(n_estimators=100, random_state=42)
rf_h.fit(X_train_h, y_train_h)
```

```python
# Predict on test data
y_pred_h = rf_h.predict(X_test_h)
```

```python
# Calculate metrics
mse_h = mean_squared_error(y_test_h, y_pred_h)
r2_h = r2_score(y_test_h, y_pred_h)
mae_h = mean_absolute_error(y_test_h, y_pred_h)

print(f"Mean Squared Error: {mse_h:.2f}")
print(f"R-squared Score: {r2_h:.2f}")
print(f"Mean absolute error: {mae_h: .2f}")
```

```python
# Hourly data with randomizedSearchcv
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    'n_estimators': [200, 300, 500],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2']
}

rf = RandomForestRegressor(random_state=42)
```

```python
rf_random = RandomizedSearchCV(
    rf,
    param_distributions=param_dist,
    n_iter=20,
    cv=3,
    scoring='r2',
    n_jobs=-1,
    random_state=42
)
```

```python
rf_random.fit(X_train_h, y_train_h)

best_rf = rf_random.best_estimator_
```

```python
y_pred_tuned = best_rf.predict(X_test_h)
mae_h_tuned = mean_absolute_error(y_test_h, y_pred_tuned)

print("Tuned R²:", r2_score(y_test_h, y_pred_tuned))
print("Tuned mae:",mean_absolute_error(y_test_h, y_pred_tuned))
```

### Test & Train accuracy

```python
def evaluate_model(model, X_train, y_train, X_test, y_test, name):
    #  Predict on Training Data
    y_train_pred = model.predict(X_train)
    #  Predict on Testing Data
    y_test_pred = model.predict(X_test)

#  Calculate R2 Scores (Accuracy)
    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)
    
#  Calculate RMSE (Error)
    train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
    
    print(f"--- {name} Data Performance ---")
    print(f"Training R2 Score: {train_r2:.4f} (Accuracy on seen data)")
    print(f"Testing  R2 Score: {test_r2:.4f} (Accuracy on unseen data)")
    print(f"Gap (Overfitting): {train_r2 - test_r2:.4f}")
    print(f"Training RMSE:     {train_rmse:.2f}")
    print(f"Testing  RMSE:     {test_rmse:.2f}")
    print("="*40)
    return train_r2, test_r2
```

```python
# --- Run for Hourly Data ---
r2_train_h, r2_test_h = evaluate_model(rf_h, X_train_h, y_train_h, X_test_h, y_test_h, "HOURLY")

# --- Run for Daily Data ---
r2_train_d, r2_test_d = evaluate_model(rf_d, X_train_d, y_train_d, X_test_d, y_test_d, "DAILY")
```

### Predicting demand for next 6 hours

```python
#  Get the last known data point (The "Current" state)
last_row_h = data_h_clean.iloc[-1]

feature_cols_h = X_train_h.columns 
```

```python
#  Generate Data for Next 6 Hours
future_data_h = []
current_hr = int(last_row_h['hr'])
current_weekday = int(last_row_h['weekday'])

for i in range(1, 7): # Next 6 hours
    next_hour = (current_hr + i) % 24
    
    # If hour rolls over 23 -> 0, it's the next day
    day_increment = (current_hr + i) // 24
    next_weekday = (current_weekday + day_increment) % 7
    
    # Create a new row based on the last known row
    new_row = last_row_h.copy()
    new_row['hr'] = next_hour
    new_row['weekday'] = next_weekday
    
    # Assume 'workingday' (0=Sun, 6=Sat are not working days)
    if next_weekday == 0 or next_weekday == 6:
        new_row['workingday'] = 0
    else:
        new_row['workingday'] = 1
        
    # Append to list
    future_data_h.append(new_row)
```

```python
# Create DataFrame
X_future_h = pd.DataFrame(future_data_h)
X_future_h = X_future_h[feature_cols_h]
```

```python
#  Predict using the trained model 
future_preds_h = rf_h.predict(X_future_h)
```

```python
# Display Results
print("--- PREDICTED DEMAND: NEXT 6 HOURS ---")
for i, pred in enumerate(future_preds_h, 1):
    hr = X_future_h.iloc[i-1]['hr']
    print(f"Next Hour {i} (Hour {int(hr)}:00): {int(pred)} rentals")
```

### Predicting demand for next 6 days

```python
# Get the last known data point
last_row_d = data_d_clean.iloc[-1]
feature_cols_d = X_train_d.columns
```

```python
n_days = 6

last_known = data_d_clean.iloc[-1].copy()
cnt_history = list(data_d_clean['cnt'].iloc[-7:])

future_predictions = []

for _ in range(n_days):
    
    future_row = last_known.copy()

# Update lag features
    future_row['cnt_lag1'] = cnt_history[-1]
    future_row['cnt_lag7'] = cnt_history[0]
    future_row['rolling_mean_7'] = np.mean(cnt_history)
    future_row['rolling_std_7'] = np.std(cnt_history)

# Prepare input
    X_future = future_row[X_train_d.columns].to_frame().T

# Predict
    pred = rf_d_manual.predict(X_future)[0]
    future_predictions.append(int(pred))

# Update history
    cnt_history.append(pred)
    cnt_history.pop(0)

# Move to next day
    future_row['weekday'] = (future_row['weekday'] + 1) % 7
    last_known = future_row

print("\n--- PREDICTED DEMAND: NEXT 6 DAYS ---")
for i, pred in enumerate(future_predictions, 1):
    print(f" Day {i}: {pred} rentals")
```
