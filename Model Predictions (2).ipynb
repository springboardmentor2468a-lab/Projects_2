## Outlier detection for hourly data data using Boxplot

```python
# Define continuous variables to check for outliers

cols_to_check = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']

# Visualizing Outliers for Hourly Data
plt.figure(figsize=(15, 10))
for i, col in enumerate(cols_to_check):
    plt.subplot(2, 3, i + 1)
    sns.boxplot(x=data_h[col])
    plt.title(f'Hourly: {col}')
plt.tight_layout()
plt.show()
```

```python
# Function to remove outliers from datasets

def remove_outliers_iqr(df, columns):
   
    df_clean = df.copy()
    
    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Filter the data: Keep rows within bounds
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]
        
    return df_clean


target_cols = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
```

```python
# Remove outliers

data_h_clean = remove_outliers_iqr(data_h, target_cols)

# Verify Results

print(f"Original Rows: {data_h.shape[0]}")
print(f"Cleaned Rows:  {data_h_clean.shape[0]}")
print(f"Rows Dropped:  {data_h.shape[0] - data_h_clean.shape[0]}")
```

### Test & Train accuracy

```python
def evaluate_model(model, X_train, y_train, X_test, y_test, name):
    #  Predict on Training Data
    y_train_pred = model.predict(X_train)
    #  Predict on Testing Data
    y_test_pred = model.predict(X_test)

#  Calculate R2 Scores (Accuracy)
    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)
    
#  Calculate RMSE (Error)
    train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
    
    print(f"--- {name} Data Performance ---")
    print(f"Training R2 Score: {train_r2:.4f} (Accuracy on seen data)")
    print(f"Testing  R2 Score: {test_r2:.4f} (Accuracy on unseen data)")
    print(f"Gap (Overfitting): {train_r2 - test_r2:.4f}")
    print(f"Training RMSE:     {train_rmse:.2f}")
    print(f"Testing  RMSE:     {test_rmse:.2f}")
    print("="*40)
    return train_r2, test_r2
```

```python
# --- Run for Hourly Data ---
r2_train_h, r2_test_h = evaluate_model(rf_h, X_train_h, y_train_h, X_test_h, y_test_h, "HOURLY")

# --- Run for Daily Data ---
r2_train_d, r2_test_d = evaluate_model(rf_d, X_train_d, y_train_d, X_test_d, y_test_d, "DAILY")
```

### Predicting demand for next 6 hours

```python
#  Get the last known data point (The "Current" state)
last_row_h = data_h_clean.iloc[-1]

feature_cols_h = X_train_h.columns 

#  Generate Data for Next 6 Hours
future_data_h = []
current_hr = int(last_row_h['hr'])
current_weekday = int(last_row_h['weekday'])

for i in range(1, 7): # Next 6 hours
    next_hour = (current_hr + i) % 24
    
    # If hour rolls over 23 -> 0, it's the next day
    day_increment = (current_hr + i) // 24
    next_weekday = (current_weekday + day_increment) % 7
    
    # Create a new row based on the last known row
    new_row = last_row_h.copy()
    new_row['hr'] = next_hour
    new_row['weekday'] = next_weekday
    
    # Assume 'workingday' (0=Sun, 6=Sat are not working days)
    if next_weekday == 0 or next_weekday == 6:
        new_row['workingday'] = 0
    else:
        new_row['workingday'] = 1
        
    # Append to list
    future_data_h.append(new_row)

# Create DataFrame
X_future_h = pd.DataFrame(future_data_h)
X_future_h = X_future_h[feature_cols_h]
```
