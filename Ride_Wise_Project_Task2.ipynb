Removing Outliers

```python
import pandas as pd

def remove_outliers_iqr(df):
    numeric_cols = df.select_dtypes(include='number').columns
    Q1 = df[numeric_cols].quantile(0.25)
    Q3 = df[numeric_cols].quantile(0.75)
    IQR = Q3 - Q1

    df_clean = df[~(
        (df[numeric_cols] < (Q1 - 1.5 * IQR)) |
        (df[numeric_cols] > (Q3 + 1.5 * IQR))
    ).any(axis=1)]
    
    return df_clean
```

```python
hour_clean = remove_outliers_iqr(hour)
day_clean  = remove_outliers_iqr(day)

print("Hour original:", hour.shape, " → After outlier removal:", hour_clean.shape)
print("Day original:", day.shape, " → After outlier removal:", day_clean.shape)
```
