Removing Outliers

```python
def remove_outliers_iqr(df):
    numeric_cols = df.select_dtypes(include='number').columns
    Q1 = df[numeric_cols].quantile(0.25)
    Q3 = df[numeric_cols].quantile(0.75)
    IQR = Q3 - Q1

    df_clean = df[~(
        (df[numeric_cols] < (Q1 - 1.5 * IQR)) |
        (df[numeric_cols] > (Q3 + 1.5 * IQR))
    ).any(axis=1)]
    
    return df_clean
```

```python
hour_clean = remove_outliers_iqr(hour)
day_clean = remove_outliers_iqr(day)

print("Hour original:", hour.shape, " → After outlier removal:", hour_clean.shape)
print("Day original:", day.shape, " → After outlier removal:", day_clean.shape)
```

Model Training

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
```

```python
# Hour dataset
X_hour = hour.drop('cnt', axis=1)
y_hour = hour['cnt']
```

```python
Xh_train, Xh_test, yh_train, yh_test = train_test_split(
    X_hour, y_hour, test_size=0.2, random_state=42
)
```

```python
scaler = StandardScaler()

# Hour
Xh_train_scaled = scaler.fit_transform(Xh_train)
Xh_test_scaled = scaler.transform(Xh_test)
```

```python
def evaluate_model(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)

    return mae, rmse, r2
```

```python
lr = LinearRegression()
rf = RandomForestRegressor(random_state=42)
gb = GradientBoostingRegressor(random_state=42)
models = [lr, rf, gb]
model_names = ["Linear Regression", "Random Forest", "Gradient Boosting"]
```

```python
print("HOUR DATASET MODEL PERFORMANCE\n")
for name, model in zip(model_names, models):
    mae, rmse, r2 = evaluate_model(model, Xh_train_scaled, Xh_test_scaled, yh_train, yh_test)
    print(f"{name}: MAE={mae:.2f}, RMSE={rmse:.2f}, R2={r2:.3f}")
```

```python
#Day Dataset
X_day = day.drop('cnt', axis=1)
y_day = day['cnt']

Xd_train, Xd_test, yd_train, yd_test = train_test_split(
    X_day, y_day, test_size=0.2, random_state=42
)

Xd_train_scaled = scaler.fit_transform(Xd_train)
Xd_test_scaled = scaler.transform(Xd_test)
```

```python
print("\nDAY DATASET MODEL PERFORMANCE\n")
for name, model in zip(model_names, models):
    mae, rmse, r2 = evaluate_model(model, Xd_train_scaled, Xd_test_scaled, yd_train, yd_test)
    print(f"{name}: MAE={mae:.2f}, RMSE={rmse:.2f}, R2={r2:.3f}")
```

Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['auto', 'sqrt', 'log2']
}
rf = RandomForestRegressor(random_state=42)
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,
    n_jobs=-1,
    verbose=2,
    scoring='r2'
)
grid_search.fit(Xh_train_scaled, yh_train)
print("Best Parameters:", grid_search.best_params_)

```

```python
best_rf = grid_search.best_estimator_

pred = best_rf.predict(Xh_test_scaled)

mae = mean_absolute_error(yh_test, pred)
rmse = np.sqrt(mean_squared_error(yh_test, pred))
r2 = r2_score(yh_test, pred)

print("\n🎯 Tuned Random Forest Performance on Hour Dataset:")
print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²: {r2:.3f}")
```
