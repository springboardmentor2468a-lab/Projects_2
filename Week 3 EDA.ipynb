## Dealing with outliers for daily data

```python
num_cols_d = data_d.select_dtypes(include=['int64', 'float64']).columns
num_cols_d
```

```python
def remove_outliers_iqr(df, columns):
    Q1 = df[columns].quantile(0.25)
    Q3 = df[columns].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    df_clean = df[~((df[columns] < lower_bound) | (df[columns] > upper_bound)).any(axis=1)]
    return df_clean
```

```python
data_d_clean = remove_outliers_iqr(data_d, num_cols_d)
```

```python
print("Before:", data_d.shape)
print("After :", data_d_clean.shape)
```
