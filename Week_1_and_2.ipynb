```python
import pandas as pd
import numpy as np

try:
    from google.colab import files
    _colab = True
except Exception:
    files = None
    _colab = False
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
import seaborn as sns
import matplotlib.pyplot as plt

# Load data (use relative paths so the notebook works locally and in Colab)
print("LOADING DATA...")
day_df = pd.read_csv('day.csv')
hour_df = pd.read_csv('hour.csv')

def analyze_and_clean_duplicates(df, name):
    print(f"\n{'='*20} ANALYSIS FOR: {name} {'='*20}")

    print(f"\n--- First 5 Rows ({name}) ---")
    display(df.head())

    print(f"\n--- Info ({name}) ---")
    df.info()

    print(f"\n--- Missing Values ({name}) ---")
    print(df.isnull().sum())

    print(f"\n--- Statistics ({name}) ---")
    display(df.describe())

    # Drop Duplicates
    initial_count = len(df)
    df = df.drop_duplicates()
    new_count = len(df)
    if initial_count != new_count:
        print(f"\n Duplicates Removed: {initial_count - new_count} rows dropped.")
    else:
        print("\n No duplicates found.")

    num_features = df.select_dtypes(include=[np.number]).columns.tolist()
    cat_features = df.select_dtypes(exclude=[np.number]).columns.tolist()

    print(f"\nNumerical Features: {num_features}")
    print(f"Categorical Features: {cat_features}")

    return df
# call both functions to inspect raw data
day_df = analyze_and_clean_duplicates(day_df, "Day Dataset")
hour_df = analyze_and_clean_duplicates(hour_df, "Hour Dataset")

print(f"\n{'='*20} APPLYING TRANSFORMATION {'='*20}")

new_names = {
    'yr': 'year',
    'mnth': 'month',
    'hum': 'humidity',
    'cnt': 'total_count',
    'weathersit': 'weather_condition',
    'hr': 'hour',
}
season_names = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weather_names = {1: 'Clear', 2: 'Mist', 3: 'Light Snow', 4: 'Heavy Rain'}

def clean_transform(df):
    df = df.copy()
    # Rename columns to friendly names if present
    df.rename(columns=new_names, inplace=True)

    # Date conversion
    if 'dteday' in df.columns:
        df['dteday'] = pd.to_datetime(df['dteday'])

    # Value Mapping
    if 'season' in df.columns:
        df['season'] = df['season'].map(season_names)
    if 'weather_condition' in df.columns:
        df['weather_condition'] = df['weather_condition'].map(weather_names)

    return df

day_cleaned = clean_transform(day_df)
hour_cleaned = clean_transform(hour_df)

print("\n✅ Transformation Complete!")
print("Cleaned Day Data Head:")
display(day_cleaned.head())

# Save cleaned data locally
day_cleaned.to_csv('day_cleaned.csv', index=False)
hour_cleaned.to_csv('hour_cleaned.csv', index=False)

# If running in Colab, trigger downloads; otherwise skip
if _colab and files is not None:
    try:
        files.download('day_cleaned.csv')
        files.download('hour_cleaned.csv')
    except Exception as e:
        print("Could not download files (Colab environment may have changed):", e)

# --- EDA cells merged from Week 1 & 2 Updated (adapted to cleaned names) ---

# Display cleaned data heads
print('Day cleaned head:')
display(day_cleaned.head())
print('Hour cleaned head:')
display(hour_cleaned.head())

# Basic info and duplicate checks
day_cleaned.info()
hour_cleaned.info()
print('Day duplicates:', day_cleaned.duplicated().sum())
print('Hour duplicates:', hour_cleaned.duplicated().sum())

# Ensure datetime column is proper (safe-guard)
if 'dteday' in day_cleaned.columns:
    day_cleaned['dteday'] = pd.to_datetime(day_cleaned['dteday'])
if 'dteday' in hour_cleaned.columns:
    hour_cleaned['dteday'] = pd.to_datetime(hour_cleaned['dteday'])

# Drop 'instant' if present (same as updated notebook)
if 'instant' in day_cleaned.columns:
    day_cleaned = day_cleaned.drop(columns=['instant'])
if 'instant' in hour_cleaned.columns:
    hour_cleaned = hour_cleaned.drop(columns=['instant'])

print('Columns (day):', list(day_cleaned.columns))
print('Columns (hour):', list(hour_cleaned.columns))

# Distribution of total bike rentals
plt.figure(figsize=(7,5))
sns.histplot(day_cleaned['total_count'], kde=True)
plt.title("Distribution of Total Bike Rentals")
plt.show()


# Monthly rental count (using renamed 'month')
monthly_counts = day_cleaned.groupby('month')['total_count'].sum().reset_index()
monthly_counts.style.background_gradient(subset=['total_count'], cmap='YlGnBu')

# Bike rentals over time (daily)
plt.figure(figsize=(12,5))
plt.plot(day_cleaned['dteday'], day_cleaned['total_count'])
plt.title("Bike Rentals Over Time")
plt.xlabel("Date")
plt.ylabel("Count")
plt.show()

# Rolling mean (7-day/30-day) to smooth
plt.figure(figsize=(12,4))
plt.plot(day_cleaned['dteday'], day_cleaned['total_count'].rolling(7).mean(), label='7-day MA')
plt.plot(day_cleaned['dteday'], day_cleaned['total_count'].rolling(30).mean(), label='30-day MA')
plt.legend()
plt.title("Bike Rentals with Rolling Means")
plt.show()

# Average rentals by season
plt.figure(figsize=(7,5))
sns.barplot(x='season', y='total_count', data=day_cleaned)
plt.title("Average Rentals by Season")
plt.show()

```

```python
# Convert normalized temperature to Celsius and replot (assumption: normalization by 41°C)
if 'temp' in day_cleaned.columns:
    # Assumption note: in the UCI Bike Sharing dataset 'temp' is normalized by 41 (max temp)
    SCALE_C = 41.0
    day_cleaned['temp_c'] = day_cleaned['temp'] * SCALE_C
    if 'temp' in hour_cleaned.columns:
        hour_cleaned['temp_c'] = hour_cleaned['temp'] * SCALE_C

    # Print quick comparison stats
    print('Normalized temp stats:')
    print(day_cleaned['temp'].describe())
    print('\nConverted temperature (°C) stats:')
    print(day_cleaned['temp_c'].describe())

    # Scatter: Temperature (°C) vs Rentals
    plt.figure(figsize=(7,5))
    sns.scatterplot(x='temp_c', y='total_count', data=day_cleaned, alpha=0.6)
    plt.xlabel('Temperature (°C)')
    plt.ylabel('Total rentals')
    plt.title('Temperature (°C) vs Rentals')
    plt.grid(True, linestyle='--', alpha=0.3)
    plt.show()

    # Histogram of temperature in Celsius to inspect range
    plt.figure(figsize=(7,4))
    sns.histplot(day_cleaned['temp_c'], kde=True)
    plt.xlabel('Temperature (°C)')
    plt.title('Distribution of Temperature (°C)')
    plt.show()
else:
    print("Column 'temp' not found in day_cleaned. Cannot convert to °C.")
```
