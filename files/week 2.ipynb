```python

numeric_cols = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
def remove_outliers(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df = df[(df[col] >= lower) & (df[col] <= upper)]
    return
```

```python
# Apply outlier removal

# Correct the variable names for the dataframes and numeric columns
day_before = dfd.shape[0]
hour_before = dfh.shape[0]

# Redefine the remove_outliers function to correctly return the dataframe
# as the original function in cell H9LQ4L6Qv-yg does not return the modified df
def remove_outliers_corrected(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df = df[(df[col] >= lower) & (df[col] <= upper)]
    return df
dfd = remove_outliers_corrected(dfd, numeric_cols)
dfh = remove_outliers_corrected(dfh, numeric_cols)

day_after = dfd.shape[0]
hour_after = dfh.shape[0]

print("\nDay dataset rows before:", day_before, "| after:", day_after)
print("Hour dataset rows before:", hour_before, "| after:", hour_after)

# Boxplot after removal
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.boxplot(data=dfd[numeric_cols])
plt.title("Day Dataset - After Outlier Removal")

plt.subplot(1,2,2)
sns.boxplot(data=dfh[numeric_cols])
plt.title("Hour Dataset - After Outlier Removal")

plt.tight_layout()
plt.show()

print("\nOutlier removal completed.")
```
