```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
day = pd.read_csv("day_cleaned.csv")
hour = pd.read_csv("hour_cleaned.csv")

day.head(), hour.head()
```

```python
# Drop only the date column
if 'dteday' in day.columns:
    day.drop('dteday', axis=1, inplace=True)

if 'dteday' in hour.columns:
    hour.drop('dteday', axis=1, inplace=True)
```

```python
day.info()
hour.info()
```

```python
# DAY dataset
X_day = day.drop("cnt", axis=1)
y_day = day["cnt"]

# HOUR dataset
X_hour = hour.drop("cnt", axis=1)
y_hour = hour["cnt"]
```

```python
X_day_train, X_day_test, y_day_train, y_day_test = train_test_split(
    X_day, y_day, test_size=0.2, random_state=42
)

X_hour_train, X_hour_test, y_hour_train, y_hour_test = train_test_split(
    X_hour, y_hour, test_size=0.2, random_state=42
)
```

```python
model_day = LinearRegression()
model_day.fit(X_day_train, y_day_train)

pred_day = model_day.predict(X_day_test)

print("DAY DATASET RESULTS:")
print("RMSE:", (mean_squared_error(y_day_test, pred_day))**0.5)
print("R2 Score:", r2_score(y_day_test, pred_day))
```

```python
model_hour = RandomForestRegressor(random_state=42)
model_hour.fit(X_hour_train, y_hour_train)

pred_hour = model_hour.predict(X_hour_test)

print("\nHOUR DATASET RESULTS:")
print("RMSE:", (mean_squared_error(y_hour_test, pred_hour))**0.5)
print("R2 Score:", r2_score(y_hour_test, pred_hour))
```

```python
print("\n----- MODEL COMPARISON -----")
print("Day Model (Linear Regression) R2:", r2_score(y_day_test, pred_day))
print("Hour Model (Random Forest)    R2:", r2_score(y_hour_test, pred_hour))
```
