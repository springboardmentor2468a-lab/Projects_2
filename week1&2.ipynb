```python
columns_to_drop = [
    'dteday',
    'casual',
    'registered',
    'yr',
    'instant'
]

dfd = dfd.drop(columns=columns_to_drop, errors='ignore')
dfd
```

```python

# Select only numeric columns
numeric_cols = dfd.select_dtypes(include=['int64', 'float64']).columns

def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Remove rows where values are outside bounds
    cleaned_data = data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]
    return cleaned_data

# Apply outlier removal on each numeric column
dfd_cleaned = dfd.copy()

for col in numeric_cols:
    dfd_cleaned = remove_outliers_iqr(dfd_cleaned, col)

print("Original shape:", dfd.shape)
print("Cleaned shape:", dfd_cleaned.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in numeric_cols:
    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    sns.boxplot(x=dfd[col], color='orange')
    plt.title(f"Before Cleaning: {col}")

    plt.subplot(1,2,2)
    sns.boxplot(x=dfd_cleaned[col], color='green')
    plt.title(f"After Cleaning: {col}")

    plt.show()
```
