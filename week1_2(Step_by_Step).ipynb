```python
import pandas as pd
import numpy as np

# For downloading in colab
try:
    from google.colab import files
    _colab = True
except:
    files = None
    _colab = False

import seaborn as sns
import matplotlib.pyplot as plt
```

Cell 2 — Load Datasets

```python
print("LOADING DATA...")
day_df = pd.read_csv('day.csv')
hour_df = pd.read_csv('hour.csv')
```

Cell 3 — Duplicate Analysis Function

```python
def analyze_and_clean_duplicates(df, name):

    print(f"\n{'='*20} ANALYSIS FOR: {name} {'='*20}")

    print(f"\n--- First 5 Rows ({name}) ---")
    display(df.head())

    print(f"\n--- Info ({name}) ---")
    df.info()

    print(f"\n--- Missing Values ({name}) ---")
    print(df.isnull().sum())

    print(f"\n--- Statistics ({name}) ---")
    display(df.describe())

    # Drop Duplicates
    initial_count = len(df)
    df = df.drop_duplicates()
    new_count = len(df)

    if initial_count != new_count:
        print(f"\n Duplicates Removed: {initial_count - new_count} rows dropped.")
    else:
        print("\n No duplicates found.")

    # Feature Types
    num_features = df.select_dtypes(include=[np.number]).columns.tolist()
    cat_features = df.select_dtypes(exclude=[np.number]).columns.tolist()

    print(f"\nNumerical Features: {num_features}")
    print(f"Categorical Features: {cat_features}")

    return df
```

Cell 4 — Run Duplicate Cleaning

```python
day_df = analyze_and_clean_duplicates(day_df, "Day Dataset")
hour_df = analyze_and_clean_duplicates(hour_df, "Hour Dataset")
```

Cell 5 — Rename + Transform Functions

```python
new_names = {
    'yr': 'year',
    'mnth': 'month',
    'hum': 'humidity',
    'cnt': 'total_count',
    'weathersit': 'weather_condition',
    'hr': 'hour',
}

season_names = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weather_names = {1: 'Clear', 2: 'Mist', 3: 'Light Snow', 4: 'Heavy Rain'}

def clean_transform(df):
    df = df.copy()

    df.rename(columns=new_names, inplace=True)

    if 'dteday' in df.columns:
        df['dteday'] = pd.to_datetime(df['dteday'])

    if 'season' in df.columns:
        df['season'] = df['season'].map(season_names)

    if 'weather_condition' in df.columns:
        df['weather_condition'] = df['weather_condition'].map(weather_names)

    return df
```

Cell 6 — Apply Transformation

```python
day_cleaned = clean_transform(day_df)
hour_cleaned = clean_transform(hour_df)

print("Transformation Complete!")
display(day_cleaned.head())
display(hour_cleaned.head())
```

Cell 7 — Drop Unwanted Columns

```python
columns_to_drop = ['instant', 'dteday', 'year', 'casual', 'registered']

def drop_columns_safe(df, cols):
    return df.drop(columns=[c for c in cols if c in df.columns])

day_cleaned = drop_columns_safe(day_cleaned, columns_to_drop)
hour_cleaned = drop_columns_safe(hour_cleaned, columns_to_drop)

print("Day:", list(day_cleaned.columns))
print("Hour:", list(hour_cleaned.columns))
```

Cell 14 Apply outlier Removal

```python
numeric_day = day_cleaned.select_dtypes(include=[np.number]).columns.tolist()
numeric_hour = hour_cleaned.select_dtypes(include=[np.number]).columns.tolist()

print("Numeric columns in Day:", numeric_day)
print("Numeric columns in Hour:", numeric_hour)
```

```python
def remove_outliers(df, numeric_cols):
    df_clean = df.copy()

    for col in numeric_cols:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]

    return df_clean
```

```python
day_before = day_cleaned.shape[0]
hour_before = hour_cleaned.shape[0]

day_cleaned = remove_outliers(day_cleaned, numeric_day)
hour_cleaned = remove_outliers(hour_cleaned, numeric_hour)

day_after = day_cleaned.shape[0]
hour_after = hour_cleaned.shape[0]

print("\nDay dataset rows before:", day_before, "| after:", day_after)
print("Hour dataset rows before:", hour_before, "| after:", hour_after)
```

Cell 15 MODEL TRAINING WEEK 3

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
from sklearn.preprocessing import OneHotEncoder
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

Cell 16 Hour dataset *model*

```python
X_hour = hour_cleaned.drop("total_count", axis=1)
y_hour = hour_cleaned["total_count"]

# Convert categorical → numeric
X_hour = pd.get_dummies(X_hour, drop_first=True)

print("Hour dataset shape after encoding:", X_hour.shape)
```

```python

X_train_hr, X_test_hr, y_train_hr, y_test_hr = train_test_split(
    X_hour, y_hour, test_size=0.2, random_state=42
)

print("Hour Train shape:", X_train_hr.shape)
print("Hour Test shape:", X_test_hr.shape)
```

```python

scaler_hr = StandardScaler()

X_train_hr_scaled = scaler_hr.fit_transform(X_train_hr)
X_test_hr_scaled = scaler_hr.transform(X_test_hr)
```

```python


lr_hour = LinearRegression()
lr_hour.fit(X_train_hr_scaled, y_train_hr)

hr_lr_pred = lr_hour.predict(X_test_hr_scaled)

print("\nHOUR DATASET — Linear Regression")
print("RMSE:", np.sqrt(mean_squared_error(y_test_hr, hr_lr_pred)))
print("MAE :", mean_absolute_error(y_test_hr, hr_lr_pred))
print("R² :", r2_score(y_test_hr, hr_lr_pred))
```

```python
# Random Forest
rf_hour = RandomForestRegressor(n_estimators=200, random_state=42)
rf_hour.fit(X_train_hr, y_train_hr)
hr_rf_pred = rf_hour.predict(X_test_hr)

print("\nHOUR DATASET — Random Forest")
print("RMSE:", np.sqrt(mean_squared_error(y_test_hr, hr_rf_pred)))
print("MAE :", mean_absolute_error(y_test_hr, hr_rf_pred))
print("R² :", r2_score(y_test_hr, hr_rf_pred))
```

```python
# Plot first 200 values
plt.figure(figsize=(10, 5))
plt.plot(y_test_hr.values[:200], label='Actual', linewidth=2)
plt.plot(hr_rf_pred[:200], label='Predicted (Random Forest)', linewidth=2)
plt.title("Hour Dataset (First 200 Samples): Actual vs Predicted")
plt.xlabel("Sample Index")
plt.ylabel("total_count")
plt.legend()
plt.show()
```

cell-16 Day dataset model

```python
# DAY DATASET


# 1. Split X and y
X_day = day_cleaned.drop("total_count", axis=1)
y_day = day_cleaned["total_count"]

# 2. One-hot encoding
X_day = pd.get_dummies(X_day, drop_first=True)

print("Day dataset shape after encoding:", X_day.shape)

# 3. Train-test split
X_train_day, X_test_day, y_train_day, y_test_day = train_test_split(
    X_day, y_day, test_size=0.2, random_state=42
)

print("Day Train shape:", X_train_day.shape)
print("Day Test shape:", X_test_day.shape)

# 4. Scaling
scaler_day = StandardScaler()
X_train_day_scaled = scaler_day.fit_transform(X_train_day)
X_test_day_scaled = scaler_day.transform(X_test_day)
```

```python
# Linear Regression
lr_day = LinearRegression()
lr_day.fit(X_train_day_scaled, y_train_day)
day_lr_pred = lr_day.predict(X_test_day_scaled)

print("\nDAY DATASET — Linear Regression")
print("RMSE:", np.sqrt(mean_squared_error(y_test_day, day_lr_pred)))
print("MAE :", mean_absolute_error(y_test_day, day_lr_pred))
print("R² :", r2_score(y_test_day, day_lr_pred))
```

```python
# Random Forest
rf_day = RandomForestRegressor(n_estimators=200, random_state=42)
rf_day.fit(X_train_day, y_train_day)
day_rf_pred = rf_day.predict(X_test_day)

print("\nDAY DATASET — Random Forest")
print("RMSE:", np.sqrt(mean_squared_error(y_test_day, day_rf_pred)))
print("MAE :", mean_absolute_error(y_test_day, day_rf_pred))
print("R² :", r2_score(y_test_day, day_rf_pred))
```

```python
plt.figure(figsize=(10,5))
plt.plot(y_test_day.values, label="Actual", linewidth=2)
plt.plot(day_rf_pred, label="Predicted (Random Forest)", linewidth=2)
plt.title("Day Dataset: Actual vs Predicted")
plt.xlabel("Sample Index")
plt.ylabel("total_count")
plt.legend()
plt.show()
```
