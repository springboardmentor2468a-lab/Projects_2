```python
before = hour.copy()      
num_cols = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
for col in num_cols:
    Q1 = hour[col].quantile(0.25)
    Q3 = hour[col].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
hour = hour[(hour[col] >= lower) & (hour[col] <= upper)]
print("Cleaning complete. Final rows:", hour.shape[0])
after = hour.copy()      
```

```python
print("After removing outliers:", after.shape)
print("Rows removed:", before.shape[0] - after.shape[0])
```

```python
for col in num_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=after[col])
    plt.title(f"After Outlier Removal — {col}")
    plt.show()
```
